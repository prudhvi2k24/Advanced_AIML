### Key Insights from EDA & Data Cleaning

1. **Education is highly correlated with Income:** The boxplot shows individuals earning `>50K` possess a significantly higher median number of educational years (around 12-13 years) compared to those earning `<=50K` (around 9-10 years).
2. **Gender disparity in work hours and income:** On average, males work more hours per week than females, and those earning `>50K` in both gender groups work more hours compared to those earning `<=50K`.
3. **Capped distributions prevent skewness:** Outliers in `age` and `hours-per-week` were successfully capped using IQR Winsorization, resulting in a cleaner distribution for future model training.
4. **Weak linear correlation among numerical features:** Aside from our engineered interactive features, the raw numerical variables show weak linear correlation with one another, suggesting complex non-linear structures or categorical dependencies rule the dataset.
5. **Engineered Features add valuable context:** Our newly created interactive columns—`net-capital-gain` (synthesizing market profit and loss) and `age_education_interaction` (representing the cumulative effect of experience and formal study)—provide highly descriptive indicators for predictive modelling.
